```python
%load_ext autoreload
%autoreload 2

import sys
from pathlib import Path

# Resolve project root and add to sys.path
project_root = Path.cwd()
while not (project_root / "src").exists() and project_root.parent != project_root:
    project_root = project_root.parent

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
if str(project_root / "src") not in sys.path:
    sys.path.append(str(project_root / "src"))
```

#### Bull market scenario

```python
from src.agents.tests.agent_views_test import AgentUnitTest

sut = AgentUnitTest()

bull_test_result = sut.run_agent_views_test(scenario="bullish_nvda", mocked_news=True)

if bull_test_result.report is None:
    raise ValueError("No report has been generated.")

print(f"Market sentiment: {bull_test_result.report.market_sentiment_summary}")
print(f"Number of views generated: {len(bull_test_result.report.views)}")
for i, view in enumerate(bull_test_result.report.views):
    relative_view_hypothesis = f"in relation to {view.target_asset if view.view_type == 'RELATIVE' else ""}."
    print(f"View No. {i + 1}:")
    print(f"Asset {view.base_asset} is going to change by {view.expected_outperformance} {relative_view_hypothesis}. (Confidence: {view.confidence:.2f})")
    print(f"Justification: {view.rationale}")
    print("-" * 70)
```

```python
print(bull_test_result.test_result_df.to_string())
print("=" * 70)
    
```
